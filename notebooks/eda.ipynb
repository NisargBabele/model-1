```python
plt.figure(figsize=(15, 12))

for i, column in enumerate(dfc.columns, 1):
    plt.subplot(4, 4, i)   # Adjust rows and columns as needed
    sns.boxplot(y=dfc[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python
numerical_columns = dfc.select_dtypes(include="number").columns

for column in numerical_columns:
    Q1 = dfc[column].quantile(0.25)
    Q3 = dfc[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    dfc = dfc[
        (dfc[column] >= lower) &
        (dfc[column] <= upper)
    ]
```
